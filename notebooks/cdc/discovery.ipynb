**category** and **tag**, then drill into datasets — without knowing dataset ids
up front. Backed by the Socrata catalog endpoints (`domain_categories`,
`domain_tags`, and keyword/category search), exposed as the `cdc_categories`,
`cdc_tags`, and `cdc_discover` MCP tools. The MCP server must be running.

## Browse a category's datasets

Pass a `category` (no keyword) to list what's in it.

```python
await discover(category="National Center for Health Statistics", limit=15)
```

## Keyword search within a topic

Combine a keyword with a category to narrow in.

```python
await discover("mortality", category="National Center for Health Statistics", limit=10)
```
